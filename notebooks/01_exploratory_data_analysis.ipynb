```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv('../dataset/fifa.csv')
print(f'Shape: {df.shape}')
df['primary_position'] = df['player_positions'].apply(lambda x: str(x).split(',')[0].strip())
df[['short_name', 'primary_position', 'overall', 'potential', 'club_name', 'nationality']].head(5)
```

## 1. Dataset Shape, Columns, and Missing Value Analysis
- **Total records:** 16,155
- **Total columns:** 81
- `release_clause_eur` and `mentality_composure` are 100% missing.
- All 29 in-game technical, movement, power, defending, and mentality attributes are 100% complete across all 16,155 players.
- Outfield card stats (`pace`, `shooting`, `passing`, etc.) are missing exclusively for Goalkeepers (1,775 players).
- Goalkeeper attributes (`gk_diving`, `gk_handling`, etc.) are missing exclusively for outfield players (14,380 players).

## 3. Position-Wise Attribute Profiles & Tactical Clustering
Examining mean technical ratings reveals distinctive tactical signatures:
1. **Goalkeepers:** High `goalkeeping_reflexes` (~66), extremely low outfield skills (<15).
2. **Centre-Backs (CB):** Dominant `defending_standing_tackle` (~66.2) and `power_strength` (~76.4).
3. **Wingers (LW/RW):** Superior `movement_sprint_speed` (~81) and `skill_dribbling` (~72).
4. **Playmakers (CAM/CM):** Top `attacking_short_passing` (~71) and `mentality_vision` (~68).
