## 1. Analysis of Model Results & Tactical Implications
- **Highest Raw Accuracy:** Random Forest (66.05%), demonstrating strong ensemble partition power on primary position clusters (CB, ST, GK, CM).
- **Best Macro F1 (Selected Champion):** Logistic Regression (Macro F1 = 0.4399, Weighted F1 = 0.6539), due to effective linear hyperplane separation and balanced class reweighting on rare tactical roles like LWB and RWB.
- **Lowest Log Loss:** Gradient Boosting (0.9535), offering well-calibrated class probability estimates for tactical decision support.
- **Overall ROC-AUC (OvR):** All three models exceed 0.904, proving exceptional discriminatory capability across all 15 position classes.
